# US Person Generator Example

This notebook runs the example US person PGM against the checked-in dummy count tables. The dummy assets mirror the real data layout without publishing the real data values.

## Paths

By default, the notebook uses the checked-in dummy assets. Set `US_PERSON_DATA_PATH` and `US_PERSON_AREA_CODE_DATA_PATH` to run against another local data bundle.

```python
import os
import sys
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "pyproject.toml").exists() and (
            candidate / "src" / "pgms"
        ).exists():
            return candidate
    raise RuntimeError("Could not find the sdg-pgms repository root.")


repo_root = find_repo_root(Path.cwd())
sys.path.insert(0, str(repo_root))
sys.path.insert(0, str(repo_root / "src"))

dummy_data_dir = repo_root / "examples" / "us_person" / "data"
person_data_path = Path(
    os.environ.get("US_PERSON_DATA_PATH", dummy_data_dir / "person_data_v4.tar")
)
area_code_data_path = Path(
    os.environ.get(
        "US_PERSON_AREA_CODE_DATA_PATH",
        dummy_data_dir / "zip_area_code_map.parquet",
    )
)

os.environ["US_PERSON_AREA_CODE_DATA_PATH"] = str(area_code_data_path)

person_data_path, area_code_data_path
```

## Build The PGM

```python
from examples.us_person import USPersonGenerator

if not person_data_path.exists():
    raise FileNotFoundError(person_data_path)
if not area_code_data_path.exists():
    raise FileNotFoundError(area_code_data_path)

us_person_gen = USPersonGenerator(data_path=str(person_data_path))
```

## Generate Samples

```python
samples = us_person_gen.generate_samples(
    size=25,
    seed=0,
    disable_progress_bar=True,
)
samples.head()
```
