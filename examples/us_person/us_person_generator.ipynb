# US Person Generator Example

This notebook runs the example US person PGM against the checked-in dummy count tables. The bundled counts are intentionally synthetic and illustrative only; samples will not reflect real US demographics. See [`README.md`](README.md) for guidance on adapting the example to other countries.

## Paths

The notebook uses the checked-in dummy assets at `examples/us_person/data/`.

```python
import sys
from pathlib import Path


def find_repo_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "pyproject.toml").exists() and (
            candidate / "src" / "pgms"
        ).exists():
            return candidate
    raise RuntimeError("Could not find the sdg-pgms repository root.")


repo_root = find_repo_root(Path.cwd())
sys.path.insert(0, str(repo_root))

dummy_data_dir = repo_root / "examples" / "us_person" / "data"
person_data_path = dummy_data_dir / "person_data.tar"
person_data_path
```

## Build The PGM

```python
from examples.us_person import USPersonGenerator

if not person_data_path.exists():
    raise FileNotFoundError(person_data_path)

us_person_gen = USPersonGenerator(data_path=str(person_data_path))
```

## Generate Samples

```python
samples = us_person_gen.generate_samples(
    size=25,
    seed=0,
    disable_progress_bar=True,
)
samples.head()
```
